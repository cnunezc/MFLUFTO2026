```python
import numpy as np

scores = [24, 23, 30, 29, 17, 16, 15]

np_scores = np.array(scores)

print(np_scores)  

print(scores*2)
print(np_scores*2)

#print(np_scores) 
print(np_scores+1)  

print(type(scores))
print(type(np_scores))
```

```python
#trabajando la lista
np_scores[3:-2]
# podemos convertirlo en una lista
#np_scores_list=np_scores.tolist()
#np_scores_list[:4]
```

```python
## subsetting específicamente con numpy 
np_scores >= 18
```

```python
np_scores[np_scores>=18]
```

# 2D Numpy arrays

```python
scores_1 = [24, 23, 30, 29, 17, 16, 15]
scores_2 = [15, 26, 24, 25, 18, 30, 23]
```

```python
np_2d_scores = np.array([scores_1,scores_2])
```

```python
np_2d_scores  #como lista de listas
```

```python
np_2d_scores.shape  
```

```python
np_2d_scores[0][2]
np_2d_scores[0,2]
```

```python
np_2d_scores[:,1:3] #todas las columnas
```

```python
np_2d_scores[1,:] #todas las columnas y primera fila
```

## Loop on a numpy array

```python
for score in scores: 
    print(score)
```

```python
for score in np_2d_scores: #no funciona para 2d
    print(score)
```

```python
for val in np.nditer(np_2d_scores):
    print(val) 
```

```python
for row in np_2d_scores:
    for x in row:
        print(x)
```
