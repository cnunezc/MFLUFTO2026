## Ejercicio 1

```python
import numpy as np

# Ejercicio 1: Función para calcular el GCD
def calcular_gcd(m, n):
    while n:
        m, n = n, m % n
    return m

print(f"El GCD de 48 y 18 es: {calcular_gcd(48, 18)}")
```

```python
import pandas as pd
from sklearn.datasets import load_iris

# Ejercicio 3: Pandas
# 1 e 2. Importar y añadir nombres de columnas
iris_data = load_iris()
df = pd.DataFrame(iris_data.data, columns=['sepal length', 'sepal width', 'petal length', 'petal width'])

# 3. Crear columna de relación (ratio) entre sépalo y pétalo
df['sepal_petal_ratio'] = df['sepal length'] / df['petal length']

# 4. Añadir columna 'target' basada en la especie (Iris-Setosa suele ser el ID 0 en sklearn)
# Nota: asignamos 1 si es setosa, 0 si no.
df['target'] = (iris_data.target == 0).astype(int)

print("Primeras filas del DataFrame Iris:")
print(df.head())
```

## Ejercicio 4

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. Descargar datos ---
# Descargamos los datos de Apple
datos = yf.download('AAPL', start='2020-01-01', end='2023-01-01')

# CORRECCIÓN: Si el DataFrame tiene MultiIndex (común en nuevas versiones),
# lo aplanamos para que las columnas sean simples: 'Open', 'High', 'Adj Close', etc.
if isinstance(datos.columns, pd.MultiIndex):
    datos.columns = datos.columns.get_level_values(0)

datos.head()
```

```python
# --- 2. Cálculos de Retorno ---
# Usamos 'Adj Close' (Cierre ajustado) para el cálculo de retornos
# Si no existiera 'Adj Close', usamos 'Close'
col_precio = 'Adj Close' if 'Adj Close' in datos.columns else 'Close'

# Calcular retornos diarios simples
datos['Retorno_Diario'] = datos[col_precio].pct_change()

# Calcular retorno acumulado
# Fórmula: (1 + r1)*(1 + r2)... - 1
datos['Retorno_Acumulado'] = (1 + datos['Retorno_Diario']).cumprod() - 1

# Mostrar el resultado final
print(f"Retorno acumulado total: {datos['Retorno_Acumulado'].iloc[-1]:.2%}")
```
