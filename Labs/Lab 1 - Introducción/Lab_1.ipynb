```python
import numpy as np

# Ejercicio 1: Función para calcular el GCD
def calcular_gcd(m, n):
    while n:
        m, n = n, m % n
    return m

print(f"El GCD de 48 y 18 es: {calcular_gcd(48, 18)}")

# Ejercicio 2: NumPy
# 1. Crear arreglo aleatorio de longitud 100
arreglo = np.random.rand(100)

# 2. Ordenar el arreglo
arreglo_ordenado = np.sort(arreglo)

# 3. Calcular métricas
media = np.mean(arreglo)
mediana = np.median(arreglo)
varianza = np.var(arreglo)

print(f"\nEstadísticas NumPy:\nMedia: {media:.4f}\nMediana: {mediana:.4f}\nVarianza: {varianza:.4f}")
```

```python
import pandas as pd
from sklearn.datasets import load_iris

# Ejercicio 3: Pandas
# 1 e 2. Importar y añadir nombres de columnas
iris_data = load_iris()
df = pd.DataFrame(iris_data.data, columns=['sepal length', 'sepal width', 'petal length', 'petal width'])

# 3. Crear columna de relación (ratio) entre sépalo y pétalo
df['sepal_petal_ratio'] = df['sepal length'] / df['petal length']

# 4. Añadir columna 'target' basada en la especie (Iris-Setosa suele ser el ID 0 en sklearn)
# Nota: Según tu slide, asignamos 1 si es setosa, 0 si no.
df['target'] = (iris_data.target == 0).astype(int)

print("Primeras filas del DataFrame Iris:")
print(df.head())
```

```python
# Instalación (Descomentar si usas Colab)
# !pip install yfinance

import yfinance as yf

# --- Técnica: Importación y variables ---
activo = 'AAPL'
precio = 150.25
precios_lista = [140, 145, 150]
info_dict = {'ticker': activo, 'precios': precios_lista}

print(f"Diccionario financiero: {info_dict}")

# --- Laboratorio: Descargar datos de Apple ---
datos = yf.download('AAPL', start='2020-01-01', end='2023-01-01')

print("\nDatos descargados de AAPL (primeras filas):")
print(datos.head())
```

```python
# 1. Calcular retornos diarios simples (basado en el precio de cierre ajustado)
datos['Retorno_Diario'] = datos['Adj Close'].pct_change()

# 2. Calcular retorno acumulado
# Fórmula: (1 + r1)*(1 + r2)*... - 1
datos['Retorno_Acumulado'] = (1 + datos['Retorno_Diario']).cumprod() - 1

# 3. Mostrar resultados
print(f"\nRetorno acumulado total al final del periodo: {datos['Retorno_Acumulado'].iloc[-1]:.2%}")

# Ver últimas filas con columnas específicas
print("\nDetalle de retornos:")
print(datos[['Adj Close', 'Retorno_Diario', 'Retorno_Acumulado']].tail())

#Tip de Python: Recuerda que al usar pct_change(), la primera fila siempre será NaN porque no tiene un día anterior con el cual compararse.
# Si quieres limpiar los datos, puedes usar datos.dropna().
```
