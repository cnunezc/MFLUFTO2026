```python
#Escribe una función que devuelve el min y el max values de la list
def minmax(data):
    
    min_value = data[0] #assuming not empty 
    max_value = data[0] #assuming not empty
    
    for i in data:
        if i < min_value:
            min_value=i
        elif i > max_value:
            max_value=i
            
    return min_value,max_value

data = [2,3,1,6]
minmax(data)
```

## Control Flow en Python

```python
# IF
# Se utiliza para decidir si una determinada instrucción o bloque de instrucciones 
# se ejecutará o no.

# Esto significa que: 
# si una cierta condición es verdadera, entonces se ejecuta un bloque de instrucciones; 
# de lo contrario, no se ejecuta.

# Cómo utilizar la sentencia IF dentro de una función
def is_multiple(n,m):
    if m!=0 and n%m==0: #el resto de la división
        return True
    else:
        return False
    
is_multiple(8,2) 
```

```python
#Esta es la primera fucnión escrita en al estilo más Python
def is_multiple(n,m):
    return m!=0 and n%m==0
    
is_multiple(5,2) 
```

```python
# FOR:
# Un bucle for se utiliza para iterar (recorrer) sobre una secuencia
# (como listas, tuplas, cadenas, rangos, etc.)

# Ejemplo del uso de un bucle for para calcular la suma de los cuadrados
# de los primeros n números
sum_squares = 0
n = 5

for i in range(n+1):  
    sum_squares+=i**2 
        
print(sum_squares)

#range(5)
```

```python
# WHILE
# Una sentencia while en Python ejecuta repetidamente una instrucción (o bloque de instrucciones)
# mientras una condición dada sea verdadera.

# Ejemplo de una función que utiliza un bucle while
def is_even(k):
    while k>0:
        k=k-2
        
    return k==0
    

is_even(8)
```

```python
# O simplemente hazlo mediante recursión
def fact(n):
    if n==1 :
        return 1
    else :
        return n*fact(n-1)
    
fact(10)
```

```python
# Cómo importar un módulo de Python en tu script

# Puedes importar toda la biblioteca
import math
math.factorial(5)    # 1*2*3*4*5 = 120

# O importar solo la función que necesitas
from math import factorial
factorial(5)         # 1*2*3*4*5 = 120
```
